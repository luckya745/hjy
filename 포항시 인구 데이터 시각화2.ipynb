```python
import pandas as pd
import matplotlib.pyplot as plt
import platform

# 1. 한글 폰트 설정
system_name = platform.system()
if system_name == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif system_name == 'Darwin':  # Mac
    plt.rc('font', family='AppleGothic')
else: # Linux
    plt.rc('font', family='NanumBarunGothic')

plt.rcParams['axes.unicode_minus'] = False

# 2. 엑셀 파일 불러오기 (수정된 부분)
# 실제 엑셀 파일명과 경로를 입력해주세요.
file_path = '/content/최근 10년간 포항시 인구 변화 데이터.xlsx

# 엑셀 파일 읽기 (sheet_name을 지정하지 않으면 첫 번째 시트를 읽습니다)
df = pd.read_excel(file_path, engine='openpyxl')

# 3. 데이터 전처리
# 컬럼명에서 ' 년' 제거 (필요한 경우)
df.columns = [str(col).replace(' 년', '') for col in df.columns]

# 연도 컬럼 리스트 추출
years = [col for col in df.columns if str(col).isdigit()]

# 데이터 추출 함수 정의
def get_data(region, item):
    row = df[(df['행정구역별(읍면동)'] == region) & (df['항목'] == item)]
    return row[years].iloc[0].astype(int)

# 4. 시각화 (이전과 동일)
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
plt.subplots_adjust(hspace=0.3, wspace=0.2)

# (1) 포항시 총인구 변화
total_pop = get_data('포항시', '총인구[명]')
axes[0, 0].plot(total_pop.index, total_pop.values, marker='o', color='navy', linewidth=2)
axes[0, 0].set_title('포항시 총인구 변화 (2015-2024)')
axes[0, 0].grid(True, linestyle='--', alpha=0.5)
axes[0, 0].set_ylabel('인구 수 (명)')

# (2) 남구 vs 북구 인구 비교
nam_gu = get_data('남구', '총인구[명]')
buk_gu = get_data('북구', '총인구[명]')
axes[0, 1].plot(nam_gu.index, nam_gu.values, marker='s', label='남구', color='green')
axes[0, 1].plot(buk_gu.index, buk_gu.values, marker='^', label='북구', color='orange')
axes[0, 1].set_title('구별 인구 추이 (남구 vs 북구)')
axes[0, 1].legend()
axes[0, 1].grid(True, linestyle='--', alpha=0.5)

# (3) 남녀 인구 변화 (포항시 전체)
male = get_data('포항시', '남자[명]')
female = get_data('포항시', '여자[명]')
axes[1, 0].plot(male.index, male.values, marker='o', label='남자', color='blue')
axes[1, 0].plot(female.index, female.values, marker='o', label='여자', color='red')
axes[1, 0].set_title('포항시 남녀 인구 변화')
axes[1, 0].legend()
axes[1, 0].grid(True, linestyle='--', alpha=0.5)

# (4) 외국인 인구 변화
foreigner = get_data('포항시', '외국인-계[명]')
axes[1, 1].bar(foreigner.index, foreigner.values, color='purple', alpha=0.7)
axes[1, 1].set_title('포항시 외국인 인구 변화')
axes[1, 1].grid(True, axis='y', linestyle='--', alpha=0.5)
axes[1, 1].set_ylabel('인구 수 (명)')

plt.show()
```
